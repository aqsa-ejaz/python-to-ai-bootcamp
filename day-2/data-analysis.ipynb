# how many people survived?

```python
print("Survival counts:")

display(df["Survived"].value_counts())

print("Survival percentages:")

display(
    df["Survived"].value_counts(normalize=True) * 100
)
```

```python
# My custom question 1:
# How many female passengers in 3rd class survived?
q1 = df.loc[(df["Sex"] == "female") & (df["Pclass"] == 3) & (df["Survived"] == 1)]
print("Female survivors in 3rd class:", len(q1))

# My custom question 2:
# What is the average age of passengers who did NOT survive?
q2 = df.loc[df["Survived"] == 0, "Age"].mean()
print("Average age of non-survivors:", round(q2, 1))
```

Q1: 72 women in 3rd class survived. Q2: the average age of non-survivors was 30.6.
